```python
stock = {
    'pains au chocolat':500,
    'croissants':400,
    'baguettes':500,
    'beignets':200
}
```

```python
for i in stock.keys():
    print(f"{i} : {stock[i]}")
```

```python
def fabriquer(item,quantity,stock)
    if item not in stock.keys():
        stock[item]=quantity
        return stock
    else:
        return "Item déjà présent"
```

```python
fabriquer('mangues',100,stock)
```
